```python
import pandas as pd
import datetime
# Load product data into df. Use Latin 1 encoding to avoid issue with reading umlauts.
filename = r"C:\Users\LindseyBuss\Documents\OBETA_Project\002 product_data.csv"
column_names = ['product_id', 'product_description', 'product_group']
data_types = {
    'product_id': 'string', 
   'product_description': 'string',
   'product_group': 'category'
}
product_df = pd.read_csv(filename, names=column_names, header=None, dtype=data_types, encoding='latin-1')
print(product_df)
```

```python
def clean_products(df):
    # Remove missing values
    df = df.dropna()

    # Remove duplicates
    df = df.drop_duplicates()

    # Drop product descriptions
    df = df.drop('product_description', axis=1)

    # Truncate product group, leaving only the group number
    df['product_group'] = product_df['product_group'].apply(lambda x: x.split('_')[0])

    return df
```

```python
cleaned_product_df = clean_products(product_df)
date = str(datetime.date.today())
cleaned_product_df.to_csv(f'clean_product_data_{date}.csv', index=False)
```
