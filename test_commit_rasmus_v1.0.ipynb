```python
# Snake Game Start Screen
import pygame
import sys

pygame.init()

# Screen dimensions
WIDTH, HEIGHT = 600, 400
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption('Snake Game')

# Colors
WHITE = (255, 255, 255)
GREEN = (0, 200, 0)
BLACK = (0, 0, 0)

# Fonts
font_title = pygame.font.SysFont('Arial', 48)
font_sub = pygame.font.SysFont('Arial', 28)

# Render text
text_title = font_title.render('SNAKE GAME', True, GREEN)
text_start = font_sub.render('Press SPACE to Start', True, WHITE)

# Main loop for start screen
def start_screen():
    while True:
        screen.fill(BLACK)
        screen.blit(text_title, (WIDTH//2 - text_title.get_width()//2, HEIGHT//3))
        screen.blit(text_start, (WIDTH//2 - text_start.get_width()//2, HEIGHT//2))
        pygame.display.flip()
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_SPACE:
                    return

# To test the start screen, uncomment the following line:
# start_screen()
# print('Game Started!')
```
